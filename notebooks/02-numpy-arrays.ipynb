# NumPy Arrays -- Ch 2.3-2.4

Computation on NumPy arrays can be very fast, or it can be very slow. The key to making it fast is to use vectorized operations(convenient interface into just this kind of statically typed, compiled routine), generally implemented through NumPy's universal functions (ufuncs).

Ufuncs exist in two flavors: unary ufuncs, which operate on a single input, and binary ufuncs, which operate on two inputs. 

ARRAY ARITHMETIC

```python
import numpy as np
x = np.arange(4)
print("x      =", x)
print("x + 5  =", x + 5)
print("x - 5  =", x - 5)
print("x * 5  =", x * 5)
print("x / 2  =", x / 2)
print("x // 2 =", x // 2) #floor division
print("-x     =", -x)
print("x ** 2 =", x ** 2) #exponentiation
print("x % 2  =", x % 2) #modulus
```
