# DTYPES

```python
import pandas as pd

df = pd.DataFrame({
    "country": ["Spain", None, "France", "France", "Italy"],
    "variety": [
        "Tempranillo",
        "Garnacha",
        "Merlot",
        "Chardonnay",
        "Sangiovese"
    ],
    "points": [92, 89, 94, 91, 93],
    "price": [30, 25, 35, 40, 28]
})

df
```

```python
df.price.dtype

#int64 means a 64-bit integer number
```

```python
df.dtypes
```

```python
#trasform points column from existing int64 into float64
df.points.astype('float64')
```

```python
#dataframe or series index has its own dtype too
df.index.dtype
```

# MISSING DATA

Entries missing values are given the value NaN ("Not a number"). For technical issues, these NaN values are always of the float64 dtype.

- pd.isnull()
- pd.notnull()

```python
df[pd.isnull(df.country)]
```

```python
#for replacing missing values
df.country.fillna("Unknown")
```

```python
#update a value
df.price.replace(30, 32)
```
