MAPS

A function that takes one set of values and "maps" them to another set of values.
Two methods:

1. map()
The function you pass to map() should expect a single value from the Series (a point value, in the above example), and return a transformed version of that value. map() returns a new Series where all the values have been transformed by your function.

2. apply()
Equivalent method if we want to transform a whole DataFrame by calling a costum method on each row.

```python
#remean the temperatures to 0

df_temp_mean = df.temperature_c.mean()
df.temperature_c.map(lambda p: p - df_temp_mean)
```

```python
#if there was a description column, count how many times the words "urban" and "traffic" appeared
n_urban = df.description.map(lambda desc: "urban" in desc).sum()
n_traffic = df.description.map(lambda desc: "traffic" in desc).sum()
count = pd.Series([n_urban, n_traffic], index=['urban', 'traffic'])
```

```python
#tranform DataFrame by calling a costum method on each row.
def remean_temp(row):
    row.temperature_c = row.temperature_c - df_temp_mean
    return row

df.apply(remean_temp, axis='columns')
```

map() and apply() return new, transformed Series and DataFrames, repectively. They don't modify the original data they are called on.
