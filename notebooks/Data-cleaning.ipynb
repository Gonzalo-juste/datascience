# HANDLING MISSING VALUES

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Ana", "Ben", None, "David", "Eva", "Frank"],
    "Age": [24, np.nan, 31, 28, np.nan, 36],
    "City": ["Madrid", "Paris", "London", None, "Rome", "Berlin"],
    "Salary": [32000, 41000, np.nan, 39000, 45000, np.nan],
    "Active": [True, False, True, None, True, False],
}

df = pd.DataFrame(data)

print(df)

#both None and np.nan used for missing values
#in the 'Active' column, it is displayed as None bc it is boolean type, but still recognized as empty
```

```python
#num of missing values per column
missing_values_count = df.isnull().sum()

#display
missing_values_count[0:]
```

```python
#percentage of data missing
total_cells = np.product(df.shape)
total_missing = missing_values_count.sum()
percentage_missing = (total_missing/total_cells) * 100

percentage_missing.round(2)
```

```python
#remove rows with missing values
df.dropna()
```

```python
#remove all columns with missing values
cols_with_na = df.dropna(axis=1)
cols_with_na.head()
```

```python
#replace all NA's with 0
filled_df = df.fillna(0)
filled_df
```

```python
#replace all NA's with whatever value comes directly after in the same column.
#then replace all the remaining na's with 0
filled_df = df.bfill(axis=0).fillna(0)
filled_df
```
