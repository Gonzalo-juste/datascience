# SUMMARY FUNCTIONS AND MAPS

SUMMARY FUNCTIONS

```python
import pandas as pd

data = {
    "city": ["Edinburgh", "Valencia", "Madrid", "London", "Paris", "Berlin", "Lisbon", "Rome"],
    "country": ["UK", "Spain", "Spain", "UK", "France", "Germany", "Portugal", "Italy"],
    "temperature_c": [12, 24, 20, 15, 18, 16, 22, 21],
    "population_millions": [0.5, 0.8, 6.7, 9.0, 2.1, 3.7, 0.5, 2.8],
    "is_coastal": [False, True, False, False, False, False, True, False]
}

df = pd.DataFrame(data)
df
```

```python
#high level summary of the attributes in the given column
df.population_millions.describe()
```

```python
#mean 
df.temperature_c.mean()
```

```python
#list of unique values
df.country.unique()
```

```python
#list of unique values and how often they occur in the dataset
df.country.value_counts()
```

MAPS

A function that takes one set of values and "maps" them to another set of values.
Two methods:

1. map()
The function you pass to map() should expect a single value from the Series (a point value, in the above example), and return a transformed version of that value. map() returns a new Series where all the values have been transformed by your function.

2. apply()
Equivalent method if we want to transform a whole DataFrame by calling a costum method on each row.

```python
#remean the temperatures to 0

df_temp_mean = df.temperature_c.mean()
df.temperature_c.map(lambda p: p - df_temp_mean)
```

```python
#if there was a description column, count how many times the words "urban" and "traffic" appeared
n_urban = df.description.map(lambda desc: "urban" in desc).sum()
n_traffic = df.description.map(lambda desc: "traffic" in desc).sum()
count = pd.Series([n_urban, n_traffic], index=['urban', 'traffic'])
```

```python
#tranform DataFrame by calling a costum method on each row.
def remean_temp(row):
    row.temperature_c = row.temperature_c - df_temp_mean
    return row

df.apply(remean_temp, axis='columns')
```

map() and apply() return new, transformed Series and DataFrames, repectively. They don't modify the original data they are called on.

# GROUPING

```python
#new DataFrame
import pandas as pd

reviews = pd.DataFrame({
    "country": [
        "Spain", "Spain", "Spain",
        "France", "France", "France",
        "Italy", "Italy", "Italy",
        "Portugal", "Portugal",
        "USA", "USA", "USA", "USA"
    ],
    "variety": [
        "Tempranillo", "Garnacha", "Albariño",
        "Merlot", "Chardonnay", "Pinot Noir",
        "Sangiovese", "Nebbiolo", "Pinot Grigio",
        "Touriga Nacional", "Alvarinho",
        "Cabernet Sauvignon", "Zinfandel", "Chardonnay", "Merlot"
    ],
    "winery": [
        "Bodega Sol", "Rioja Estate", "Atlantic Wines",
        "Chateau Vert", "Maison Bleu", "Burgundy Hills",
        "Villa Rosa", "Piedmont Estate", "Veneto Wines",
        "Douro Hills", "Vinho Verde",
        "Golden Vine", "Napa Estate", "Pacific Cellars", "Sonoma Wines"
    ],
    "year": [
        2020, 2020, 2021,
        2019, 2020, 2020,
        2018, 2019, 2021,
        2020, 2021,
        2019, 2020, 2020, 2021
    ],
    "points": [
        92, 89, 91,
        90, 94, 93,
        88, 95, 87,
        91, 89,
        96, 90, 94, 92
    ],
    "price": [
        30, 30, 25,
        40, 35, 35,
        20, 50, 20,
        25, 25,
        60, 30, 30, 45
    ]
})
reviews
```

```python
#same as value_counts()
reviews.groupby('points').points.count()
```

```python
#cheapest in each point value
reviews.groupby('points').price.min()
```

```python
#name of the first variety of wine reviewed from each year in the dataset
reviews.groupby('year').apply(lambda df: df.variety.iloc[0])
```

```python
#best wine by country and price
reviews.groupby(['country', 'price']).apply(lambda df: df.loc[df.points.idxmax()])
```

```python
#agg() lets you run a bunch of different functions on your DataFrame simultaneously
reviews.groupby(['country']).price.agg([len, min, max])
```

# MULTI-INDEXES

```python
countries_reviewed = reviews.groupby(['country', 'year']).winery.agg([len])
countries_reviewed
```
